# Research Question

How do production budget, movie genre, lead actors, language, and country of production influence a film's financial and critical success — measured by box office revenue and IMDb ratings — and does this relationship differ between US and non-US productions, and across the pre- and post-streaming eras (before and after 2015)?

Specifically, we examine whether the budget-success relationship has weakened as streaming platforms have disrupted theatrical revenue models, and whether Hollywood's historically dominant global distribution network produces a stronger budget-revenue association compared to non-US film industries. We are also interested in whether lower-budget international films in certain genres can achieve comparable critical success (IMDb ratings) despite lower revenue, suggesting that financial and critical success are not always aligned.

The key variables include production budget (numerical), genre (categorical), country of production (categorical), lead actor (categorical), original language (categorical), release year (numerical, used to define pre/post-2015 periods), box office revenue (numerical), and IMDb rating (numerical). We will use correlation analysis, group comparisons, and interaction-based statistical models to examine these relationships. This is an inferential data analysis task that builds on and extends prior work by incorporating geographic, temporal, and talent-related dimensions that the existing literature has largely overlooked.

## Background and Prior Work

The film industry is strongly influenced by production budgets, which affect everything from casting and visual effects to marketing and distribution. Different movie genres tend to require different levels of funding. Research on box office revenue prediction has confirmed that production budget is the single strongest predictor of a film's financial performance, with action and adventure genres consistently commanding the highest budgets while drama and horror operate at far lower cost levels.<sup>1</sup>

Previous peer-reviewed work has also shown that genre plays a major role in determining production cost and financial outcomes. Shahid and Islam (2023) demonstrated that a genre's popularity at the time of release is a strong predictor of box office returns, and that drama films consistently carry lower budgets than action or adventure films — underscoring how budget and genre interact to shape financial outcomes.<sup>2</sup>

When it comes to critical success, the picture is different. A study analyzing IMDb user voting data found no significant correlation between production budget and average user ratings, suggesting that audience satisfaction is driven by factors such as storytelling quality and emotional resonance rather than spending alone.<sup>3</sup> This supports our hypothesis that financial and critical success are not always aligned, and that lower-budget international films may achieve high ratings despite modest revenues.

Despite this body of work, several important gaps remain. First, most existing studies treat the global film industry as a monolith and focus almost exclusively on Hollywood. However, industry data show that international revenues for US studio films now account for over 70% of Hollywood's total box office take, and that non-US industries operate under fundamentally different distribution conditions.<sup>4</sup> Cross-national comparisons of how budget, language, and talent translate to success are therefore critical but largely absent from the literature.

Second, few studies account for the structural disruption caused by streaming platforms. Research published in Applied Economics found that the launch of Netflix streaming in the US was associated with a 14–17% reduction in theatrical box office revenues, establishing a clear empirical link between streaming availability and declining theatrical performance.<sup>5</sup> This suggests that the budget-revenue relationship documented in pre-streaming studies may no longer hold post-2015, yet most prior analyses ignore this temporal shift. Our project directly addresses both gaps by incorporating country of production and original language as comparative variables, and by splitting our analysis across pre- and post-2015 time periods.

# Hypothesis

We hypothesize that movies with larger production budgets, globally recognized actors, and English-language releases will generally earn higher box office revenue due to wider international distribution and marketing reach. Specifically, we expect this budget-revenue relationship to be stronger for US films than for non-US films, because Hollywood benefits from an established global distribution and marketing infrastructure that amplifies returns on high-budget productions — a network advantage that non-US industries generally lack.

We further hypothesize that this budget-revenue association has weakened in the post-2015 period compared to the pre-2015 era, as the proliferation of streaming platforms has reduced audiences' dependence on theatrical releases, particularly for lower-budget films that would previously have relied on box office performance as their primary revenue source.

In contrast, we expect IMDb ratings to remain largely independent of production budget regardless of country of production or time period, consistent with prior empirical findings showing no significant correlation between budget and average user ratings. We also predict that some lower-budget international films, particularly in drama or thriller genres, may receive disproportionately high IMDb ratings relative to their revenue — suggesting that critical success and financial success are not always strongly aligned.

Among genres, we expect action, adventure, and fantasy films to show the strongest positive budget-revenue correlation, since these genres depend most heavily on spectacle and wide theatrical distribution to justify their investment, while drama and horror — genres that can succeed critically and commercially on low budgets — will show weaker or more variable correlations.

### Dataset #1 — Movie Industry

This dataset contains 6,820 films released between 1986 and 2016 with 15 variables covering production details, financial performance, and audience reception. The core financial variables are `budget` (production cost in USD) and `gross` (total theatrical box office revenue in USD). The `score` variable is an IMDb user rating on a 0–10 scale and serves as our primary measure of critical/audience success.

Budget values range widely — from a few thousand dollars for micro-budget films to over $300 million for major blockbusters — with a heavily right-skewed distribution. We address this by log-transforming both budget and gross after adjusting for inflation using BLS CPI-U data (base year: 2020), since raw dollar values are not comparable across four decades. A significant number of entries have missing or zero budget values (likely unreported rather than truly zero-budget). We drop these rows and document the extent of missingness below. `votes` are excluded from predictive modeling since they are a post-release variable and would introduce data leakage.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('data/00-raw/movies.csv')
print('Shape:', df.shape)
df.head()
```

```python
# Missingness summary
missing_percent = (df.isnull().mean() * 100).sort_values(ascending=False)
missing_table = pd.DataFrame({
    'Missing Values': df.isnull().sum(),
    'Percent Missing': missing_percent
})
missing_table.sort_values(by='Percent Missing', ascending=False)
```

```python
# Parse release date, drop post-release variable
df['release_date'] = pd.to_datetime(
    df['released'].str.extract(r'^([^(]+)').squeeze().str.strip(), errors='coerce'
)
df = df.drop(columns=['released', 'votes'])

str_cols = df.select_dtypes(include='object').columns
df[str_cols] = df[str_cols].apply(lambda s: s.str.strip())

for col in ['budget', 'gross', 'runtime']:
    df[col] = pd.to_numeric(df[col], errors='coerce')

df.loc[df['budget'] == 0, 'budget'] = np.nan
df.loc[df['gross'] == 0, 'gross'] = np.nan

print('After cleaning, shape:', df.shape)
print('Missing budget:', df['budget'].isnull().sum())
print('Missing gross:', df['gross'].isnull().sum())

df = df.dropna(subset=['budget'])
print('After dropping missing budget:', df.shape)
```

```python
# CPI-U inflation adjustment to 2020 dollars
cpi = {
    1980: 82.4, 1981: 90.9, 1982: 96.5, 1983: 99.6, 1984: 103.9,
    1985: 107.6, 1986: 109.6, 1987: 113.6, 1988: 118.3, 1989: 124.0,
    1990: 130.7, 1991: 136.2, 1992: 140.3, 1993: 144.5, 1994: 148.2,
    1995: 152.4, 1996: 156.9, 1997: 160.5, 1998: 163.0, 1999: 166.6,
    2000: 172.2, 2001: 177.1, 2002: 179.9, 2003: 184.0, 2004: 188.9,
    2005: 195.3, 2006: 201.6, 2007: 207.3, 2008: 215.3, 2009: 214.5,
    2010: 218.1, 2011: 224.9, 2012: 229.6, 2013: 233.0, 2014: 236.7,
    2015: 237.0, 2016: 240.0, 2017: 245.1, 2018: 251.1, 2019: 255.7,
    2020: 258.8
}
BASE_CPI = cpi[2020]
df['cpi'] = df['year'].map(cpi)
df['budget_real'] = df['budget'] * (BASE_CPI / df['cpi'])
df['gross_real'] = df['gross'] * (BASE_CPI / df['cpi'])
df = df.drop(columns=['cpi'])

df['log_budget'] = np.log1p(df['budget_real'])
df['log_gross'] = np.log1p(df['gross_real'])
df['era'] = df['year'].apply(lambda y: 'Post-2015' if y >= 2015 else 'Pre-2015')
df['us_film'] = df['country'].apply(lambda c: 'US' if c == 'United States' else 'Non-US')

print('Final Dataset 1 shape:', df.shape)
df[['budget_real','gross_real','score','runtime']].describe()
```

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 4))
df['log_budget'].hist(bins=40, ax=axes[0], color='steelblue', edgecolor='white')
axes[0].set_title('Log Budget (2020 USD)')
axes[0].set_xlabel('log(budget)')
df['log_gross'].hist(bins=40, ax=axes[1], color='coral', edgecolor='white')
axes[1].set_title('Log Gross Revenue (2020 USD)')
axes[1].set_xlabel('log(gross)')
df['score'].hist(bins=30, ax=axes[2], color='mediumseagreen', edgecolor='white')
axes[2].set_title('IMDb Score Distribution')
axes[2].set_xlabel('IMDb Score (0-10)')
plt.tight_layout()
plt.show()

print('Movies by Era:')
print(df['era'].value_counts())
print('\nUS vs Non-US:')
print(df['us_film'].value_counts())

df.reset_index(drop=True).to_csv('data/02-processed/movies_clean.csv', index=False)
print('\nDataset 1 saved.')
```

### Merging the Datasets

We merge on movie title and release year after standardizing column names. An inner join is used to produce a combined dataset with full information from both sources. Unmatched rows are noted for reference. After merging we add log-transformed financial columns and the era/US indicator columns needed for EDA.

```python
df1 = pd.read_csv('data/00-raw/movies.csv')
df2 = pd.read_csv('data/00-raw/movies_metadata.csv')
print(f'Dataset 1: {df1.shape}')
print(f'Dataset 2: {df2.shape}')

df1['merge_title'] = df1['name'].str.lower().str.strip()
df1['merge_year'] = df1['year']
df2['merge_title'] = df2['title'].str.lower().str.strip()
df2['merge_year'] = pd.to_datetime(df2['release_date'], errors='coerce').dt.year

df1_clean = df1.dropna(subset=['merge_title', 'merge_year']).copy()
df2_clean = df2.dropna(subset=['merge_title', 'merge_year']).copy()
df2_clean['merge_year'] = df2_clean['merge_year'].astype(int)

df1_cols = ['merge_title', 'merge_year', 'name', 'year', 'genre', 'rating',
            'score', 'votes', 'director', 'star', 'country', 'budget',
            'gross', 'company', 'runtime']
df2_cols = ['merge_title', 'merge_year', 'genres', 'original_language',
            'production_countries', 'popularity', 'revenue', 'vote_average',
            'vote_count', 'cast']

merged = pd.merge(
    df1_clean[df1_cols], df2_clean[df2_cols],
    on=['merge_title', 'merge_year'], how='inner', suffixes=('_ds1', '_ds2')
)
merged = merged.drop(columns=['merge_title', 'merge_year'])

# Derived columns
merged['log_budget'] = np.log1p(merged['budget'])
merged['log_gross'] = np.log1p(merged['gross'])
merged['era'] = merged['year'].apply(lambda y: 'Post-2015' if y >= 2015 else 'Pre-2015')
merged['is_us'] = merged['country'] == 'United States'

print(f'\nMerged dataset: {merged.shape}')
print(f'Columns: {merged.columns.tolist()}')
merged.to_csv('data/02-processed/merged_movies.csv', index=False)
merged.head()
```

#### Section 1: Univariate Distributions of Key Numeric Variables

We start by examining the distributions of our three core metrics: production budget, gross revenue, and IMDb score. Budget and gross are both heavily right-skewed — a small number of blockbusters earn orders of magnitude more than typical films — which is why we plot them on a log scale. The log-transformed distributions are approximately bell-shaped, confirming that a log-normal model is appropriate for these variables. IMDb score is roughly normally distributed and centered around 6.5–7.0, consistent with prior literature. The median budget ($30M) is far lower than the mean, further confirming the right skew and justifying our use of medians for group comparisons.

```python
fig, axes = plt.subplots(1, 3, figsize=(16, 4))

axes[0].hist(df['budget'].dropna(), bins=50, color='steelblue', edgecolor='white')
axes[0].set_xscale('log')
axes[0].set_xlabel('Production Budget (USD, log scale)')
axes[0].set_ylabel('Number of Movies')
axes[0].set_title(f'Budget Distribution (n={df["budget"].notna().sum()})')

axes[1].hist(df['gross'].dropna(), bins=50, color='seagreen', edgecolor='white')
axes[1].set_xscale('log')
axes[1].set_xlabel('Gross Revenue (USD, log scale)')
axes[1].set_ylabel('Number of Movies')
axes[1].set_title(f'Gross Revenue Distribution (n={df["gross"].notna().sum()})')

axes[2].hist(df['score'].dropna(), bins=30, color='coral', edgecolor='white')
axes[2].set_xlabel('IMDb Score (0-10)')
axes[2].set_ylabel('Number of Movies')
axes[2].set_title(f'IMDb Score Distribution (n={df["score"].notna().sum()})')

plt.tight_layout()
plt.show()

print(f'Median budget:  ${df["budget"].median():,.0f}')
print(f'Median gross:   ${df["gross"].median():,.0f}')
print(f'Median score:   {df["score"].median():.1f}')
```

#### Section 2: Genre Distribution

Understanding genre composition is critical before making group-level comparisons, because genres with very few films will produce unreliable statistics. Action (905) and Comedy (849) dominate our merged sample, followed by Drama (506). Genres like Thriller (4), Family (3), Romance (2), and Sci-Fi (2) are too sparse for meaningful group-level inference and should be interpreted with caution or excluded from genre comparisons. This imbalance also means that overall dataset-level statistics are heavily influenced by Action and Comedy films.

```python
genre_counts = df['genre'].value_counts()

fig, ax = plt.subplots(figsize=(10, 5))
sns.barplot(x=genre_counts.values, y=genre_counts.index, palette='viridis', ax=ax)
ax.set_xlabel('Number of Movies')
ax.set_ylabel('Genre')
ax.set_title('Movie Genre Distribution in Merged Dataset')
for i, v in enumerate(genre_counts.values):
    ax.text(v + 10, i, str(v), va='center', fontsize=9)
plt.tight_layout()
plt.show()
```

#### Section 3: Budget vs. Gross Revenue by Genre

We plot budget vs. gross on a log-log scale for the top 6 genres. The dashed line represents break-even (gross = budget). Most films cluster above this line, meaning they recoup their budget theatrically, though the wide spread confirms that financial outcomes are highly variable even at similar budget levels. The overall log-log correlation of r ≈ 0.67 confirms a strong positive relationship between budget and gross — higher-budget films tend to earn more — which is consistent with our hypothesis and prior literature. However, the substantial scatter around the trend line indicates that budget alone far from determines a film's financial outcome.

```python
top_genres = df['genre'].value_counts().head(6).index.tolist()
df_plot = df[df['genre'].isin(top_genres)].dropna(subset=['budget', 'gross'])

fig, ax = plt.subplots(figsize=(10, 6))
sns.scatterplot(data=df_plot, x='budget', y='gross', hue='genre',
                alpha=0.6, s=30, palette='tab10', ax=ax)
ax.set_xscale('log')
ax.set_yscale('log')
ax.set_xlabel('Production Budget (USD, log scale)')
ax.set_ylabel('Gross Revenue (USD, log scale)')
ax.set_title('Budget vs. Gross Revenue by Genre (Top 6 Genres)')
xlim = ax.get_xlim()
ax.plot(xlim, xlim, 'k--', alpha=0.4, label='Gross = Budget (break-even)')
ax.legend(bbox_to_anchor=(1.02, 1), loc='upper left')
plt.tight_layout()
plt.show()

log_corr = np.log10(df_plot['budget']).corr(np.log10(df_plot['gross']))
print(f'Overall log-log correlation (budget vs gross): r = {log_corr:.3f}')
```

#### Section 4: Correlation Heatmap

The pairwise correlation matrix confirms two key findings central to our research question. First, budget and gross are strongly correlated (r ≈ 0.74), validating the widely documented budget-revenue relationship. Second, and critically, budget shows essentially zero correlation with IMDb score (r ≈ 0.07). This strongly supports our hypothesis that financial and critical success are driven by fundamentally different factors — spending more does not make a film better reviewed. The weak positive correlation between gross and score (r ≈ 0.22) likely reflects that widely-seen films accumulate more votes and tend to be slightly higher quality on average, rather than a direct causal effect.

```python
corr_matrix = df[['budget', 'gross', 'score']].corr()

plt.figure(figsize=(6, 5))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt='.2f',
            linewidths=0.5, square=True)
plt.title('Correlation Matrix: Budget, Gross Revenue, and IMDb Score')
plt.tight_layout()
plt.show()
```

#### Section 5: Median Gross Revenue by Genre (Box Plot)

Box plots reveal the full distribution of gross revenue within each genre, not just the mean. Animation, Adventure, and Action show both the highest median revenues and the widest spreads — consistent with our hypothesis that these spectacle-driven genres monetize budget most effectively. Drama and Biography show much lower median revenues and tighter distributions, confirming that these genres operate in a fundamentally different financial space. Horror is notable for its wide spread: some horror films earn very little while others (think franchise entries) earn enormously — consistent with the genre's reputation for variable financial outcomes.

```python
top8 = df['genre'].value_counts().head(8).index.tolist()
df_top8 = df[df['genre'].isin(top8)].dropna(subset=['gross'])

order = df_top8.groupby('genre')['gross'].median().sort_values(ascending=False).index

fig, ax = plt.subplots(figsize=(12, 6))
sns.boxplot(data=df_top8, x='genre', y='gross', order=order,
            palette='Set2', fliersize=2, ax=ax)
ax.set_yscale('log')
ax.set_xlabel('Genre')
ax.set_ylabel('Gross Revenue (USD, log scale)')
ax.set_title('Gross Revenue Distribution by Genre (Top 8 Genres, sorted by median)')
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

#### Section 6: IMDb Score by Genre (Box Plot)

Comparing IMDb scores across genres reveals a strikingly different pattern from the revenue comparison. Drama and Biography — genres that earn relatively little at the box office — receive among the highest IMDb ratings, while Action and Animation — the top revenue earners — receive average or below-average scores. This is direct evidence supporting our hypothesis that financial success and critical/audience success are not aligned, and that genre moderates this disconnect. Horror shows notably low median scores, consistent with its reputation for variable quality.

```python
df_score = df[df['genre'].isin(top8)].dropna(subset=['score'])
score_order = df_score.groupby('genre')['score'].median().sort_values(ascending=False).index

fig, ax = plt.subplots(figsize=(12, 6))
sns.boxplot(data=df_score, x='genre', y='score', order=score_order,
            palette='Set3', fliersize=2, ax=ax)
ax.set_xlabel('Genre')
ax.set_ylabel('IMDb Score (0-10)')
ax.set_title('IMDb Score Distribution by Genre (Top 8 Genres, sorted by median)')
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()

print('Median IMDb score by genre:')
print(df_score.groupby('genre')['score'].median().sort_values(ascending=False).round(2))
```

#### Section 7: US vs. Non-US Budget-Revenue Relationship

We compare the budget-revenue relationship separately for US and non-US productions. Both groups show a positive log-linear relationship, but US films (r ≈ 0.67) have a modestly stronger correlation than non-US films (r ≈ 0.59). This is consistent with our hypothesis that Hollywood's established global distribution and marketing infrastructure amplifies returns on investment. The box plot below further shows that US films earn substantially higher median gross revenues than non-US films at comparable budget levels, even after controlling for the log scale. However, the difference in regression slopes is modest, suggesting the gap is real but not dramatic.

```python
g = sns.lmplot(data=df, x='log_budget', y='log_gross', hue='is_us',
               scatter_kws={'alpha': 0.35, 's': 15}, height=6, aspect=1.2)
plt.xlabel('Log Budget')
plt.ylabel('Log Gross Revenue')
plt.title('Budget-Revenue Relationship: US vs. Non-US Films')
plt.show()

us_corr = df[df['is_us']]['log_budget'].corr(df[df['is_us']]['log_gross'])
nonus_corr = df[~df['is_us']]['log_budget'].corr(df[~df['is_us']]['log_gross'])
print(f'US films log-log correlation:     r = {us_corr:.3f}')
print(f'Non-US films log-log correlation: r = {nonus_corr:.3f}')
```

#### Section 8: Pre/Post-2015 Streaming Era Comparison

We split the data at 2015 to test whether the budget-revenue relationship weakened after streaming became dominant. Contrary to our hypothesis, the post-2015 correlation (r ≈ 0.71) is slightly higher than pre-2015 (r ≈ 0.64). One likely explanation is survivor bias: after streaming captured lower-budget films' audiences, only high-budget wide-release films consistently appear in theatrical gross data post-2015. This would mechanically strengthen the budget-revenue correlation among films that still rely on theatrical revenue. This is an important null finding that we will explore further in subsequent analysis.

```python
g = sns.lmplot(data=df, x='log_budget', y='log_gross', hue='era',
               scatter_kws={'alpha': 0.35, 's': 15}, height=6, aspect=1.2)
plt.xlabel('Log Budget')
plt.ylabel('Log Gross Revenue')
plt.title('Budget-Revenue Relationship by Era (Pre vs. Post-2015)')
plt.show()

pre_corr = df[df['era'] == 'Pre-2015']['log_budget'].corr(df[df['era'] == 'Pre-2015']['log_gross'])
post_corr = df[df['era'] == 'Post-2015']['log_budget'].corr(df[df['era'] == 'Post-2015']['log_gross'])
print(f'Pre-2015 correlation:  r = {pre_corr:.3f}')
print(f'Post-2015 correlation: r = {post_corr:.3f}')
```

#### Section 9: Budget vs. IMDb Score

We directly test our hypothesis that IMDb score is independent of production budget. The scatter plot and near-flat regression line (r ≈ 0.00) confirm that budget has essentially no linear relationship with audience ratings across the full dataset. This is one of the clearest findings in our EDA: spending more money does not produce better-reviewed films. This decoupling of financial and critical success is consistent with prior empirical findings and supports the idea that audience satisfaction is driven by factors like storytelling, performance, and emotional resonance rather than production spending.

```python
fig, ax = plt.subplots(figsize=(8, 6))
sns.scatterplot(data=df, x='log_budget', y='score', alpha=0.4, s=15, ax=ax)
sns.regplot(data=df, x='log_budget', y='score', scatter=False, color='red',
            line_kws={'linewidth': 2}, ax=ax)
ax.set_xlabel('Log Budget')
ax.set_ylabel('IMDb Score (0-10)')
ax.set_title('Production Budget vs. IMDb Rating')
plt.tight_layout()
plt.show()

rating_corr = df['log_budget'].corr(df['score'])
print(f'Budget-IMDb correlation: r = {rating_corr:.3f}')
```

#### Section 10: Genre-Level Budget-Revenue Correlations

Computing the budget-gross correlation separately by genre confirms our directional hypothesis. Action (r ≈ 0.76), Animation (r ≈ 0.72), and Fantasy (r ≈ 0.70) show the strongest budget-revenue associations — these are precisely the spectacle-dependent genres that require large budgets and wide theatrical distribution to justify their investment. Drama (r ≈ 0.53) and Horror (r ≈ 0.37) show the weakest correlations among well-represented genres, consistent with their ability to achieve both critical and commercial success on modest budgets. The near-zero correlation for Thriller and the anomalous Western result reflect extremely small sample sizes and should not be over-interpreted.

```python
genre_corrs = (
    df.groupby('genre')
    .apply(lambda x: x['log_budget'].corr(x['log_gross']))
    .sort_values(ascending=False)
    .reset_index()
)
genre_corrs.columns = ['Genre', 'Budget-Revenue Correlation (r)']

fig, ax = plt.subplots(figsize=(10, 5))
sns.barplot(data=genre_corrs, x='Budget-Revenue Correlation (r)', y='Genre',
            palette='RdYlGn', ax=ax)
ax.axvline(0, color='black', linewidth=0.8)
ax.set_title('Budget-Revenue Correlation by Genre')
ax.set_xlabel('Pearson r (log budget vs log gross)')
plt.tight_layout()
plt.show()

print(genre_corrs.to_string(index=False))
```

#### Section 11: OLS Regression Model (Martin & Ketan)

We fit an OLS regression model predicting log gross revenue from log budget, US vs. non-US indicator, era, their interactions, genre fixed effects, and IMDb score. The model explains approximately 50% of the variance in log gross (R² ≈ 0.507). Log budget is the dominant predictor (coef ≈ 0.87, p < 0.001), confirming the budget-revenue relationship holds after controlling for all other factors. The interaction terms for US status and era are not statistically significant at conventional levels, suggesting that while raw correlations differ between groups, the budget-revenue slope does not differ significantly after controlling for genre and score. IMDb score has a small but statistically significant positive effect on gross (coef ≈ 0.20, p < 0.001), meaning better-reviewed films do earn slightly more — but the effect is far smaller than that of budget.

```python
model = smf.ols(
    """
    log_gross ~ log_budget
    + C(is_us)
    + C(era)
    + log_budget:C(is_us)
    + log_budget:C(era)
    + C(genre)
    + score
    """,
    data=df
).fit()

print(model.summary())
```
