## Abstract

This project investigates how production budget, genre, country of production, and release era influence a film's financial performance (box office gross) and critical reception (IMDb score). Using two complementary publicly available datasets covering thousands of films released between 1986 and 2017, we merged, cleaned, and analyzed records to examine these relationships across multiple dimensions. After performing exploratory data analysis, we conducted three inferential analyses: an OLS multiple regression model predicting log gross revenue, a Welch's t-test comparing the gross revenues of US versus non-US productions, and a permutation test to verify the significance of the US revenue advantage without parametric assumptions.

Our results confirm that production budget is the single strongest predictor of box office revenue, with a log-log correlation of approximately 0.67 and a statistically significant regression coefficient. Genre also plays a meaningful role — action, animation, and adventure films earn the highest revenues, while drama and biography earn far less despite receiving higher IMDb ratings. Critically, we find essentially no relationship between production budget and IMDb score (r ≈ 0.00), suggesting that financial investment does not translate into audience satisfaction. US films earn significantly higher gross revenues than non-US films (p < 0.001), consistent with Hollywood's established global distribution advantage. The budget-revenue relationship did not weaken post-2015 as hypothesized; instead, a slight strengthening likely reflects a selection effect where lower-budget films migrated to streaming while only high-budget films maintained theatrical releases. Together, these findings demonstrate that financial and critical success in film are driven by fundamentally different factors, and that no single variable fully determines a film's outcome.

## Research Question

How do production budget, movie genre, lead actors, language, and country of production influence a film's financial and critical success — measured by box office revenue and IMDb ratings — and does this relationship differ between US and non-US productions, and across the pre- and post-streaming eras (before and after 2015)?

Specifically, we examine whether the budget-success relationship has weakened as streaming platforms have disrupted theatrical revenue models, and whether Hollywood's historically dominant global distribution network produces a stronger budget-revenue association compared to non-US film industries. We are also interested in whether lower-budget international films in certain genres can achieve comparable critical success (IMDb ratings) despite lower revenue, suggesting that financial and critical success are not always aligned.

The key variables include production budget (numerical), genre (categorical), country of production (categorical), lead actor (categorical), original language (categorical), release year (numerical, used to define pre/post-2015 periods), box office revenue (numerical), and IMDb rating (numerical). We will use correlation analysis, group comparisons, and interaction-based statistical models to examine these relationships. This is an inferential data analysis task that builds on and extends prior work by incorporating geographic, temporal, and talent-related dimensions that the existing literature has largely overlooked.

## Background and Prior Work

The film industry is strongly influenced by production budgets, which affect everything from casting and visual effects to marketing and distribution. Different movie genres tend to require different levels of funding. Research on box office revenue prediction has confirmed that production budget is the single strongest predictor of a film's financial performance, with action and adventure genres consistently commanding the highest budgets while drama and horror operate at far lower cost levels.<sup>1</sup>

Previous peer-reviewed work has also shown that genre plays a major role in determining production cost and financial outcomes. Shahid and Islam (2023) demonstrated that a genre's popularity at the time of release is a strong predictor of box office returns, and that drama films consistently carry lower budgets than action or adventure films — underscoring how budget and genre interact to shape financial outcomes.<sup>2</sup>

When it comes to critical success, the picture is different. A study analyzing IMDb user voting data found no significant correlation between production budget and average user ratings, suggesting that audience satisfaction is driven by factors such as storytelling quality and emotional resonance rather than spending alone.<sup>3</sup> This supports our hypothesis that financial and critical success are not always aligned, and that lower-budget international films may achieve high ratings despite modest revenues.

Despite this body of work, several important gaps remain. First, most existing studies treat the global film industry as a monolith and focus almost exclusively on Hollywood. However, industry data show that international revenues for US studio films now account for over 70% of Hollywood's total box office take, and that non-US industries operate under fundamentally different distribution conditions.<sup>4</sup> Cross-national comparisons of how budget, language, and talent translate to success are therefore critical but largely absent from the literature.

Second, few studies account for the structural disruption caused by streaming platforms. Research published in Applied Economics found that the launch of Netflix streaming in the US was associated with a 14–17% reduction in theatrical box office revenues, establishing a clear empirical link between streaming availability and declining theatrical performance.<sup>5</sup> This suggests that the budget-revenue relationship documented in pre-streaming studies may no longer hold post-2015, yet most prior analyses ignore this temporal shift. Our project directly addresses both gaps by incorporating country of production and original language as comparative variables, and by splitting our analysis across pre- and post-2015 time periods.

## Hypothesis

We hypothesize that movies with larger production budgets, globally recognized actors, and English-language releases will generally earn higher box office revenue due to wider international distribution and marketing reach. Specifically, we expect this budget-revenue relationship to be stronger for US films than for non-US films, because Hollywood benefits from an established global distribution and marketing infrastructure that amplifies returns on high-budget productions — a network advantage that non-US industries generally lack.

We further hypothesize that this budget-revenue association has weakened in the post-2015 period compared to the pre-2015 era, as the proliferation of streaming platforms has reduced audiences' dependence on theatrical releases, particularly for lower-budget films that would previously have relied on box office performance as their primary revenue source.

In contrast, we expect IMDb ratings to remain largely independent of production budget regardless of country of production or time period, consistent with prior empirical findings showing no significant correlation between budget and average user ratings. We also predict that some lower-budget international films, particularly in drama or thriller genres, may receive disproportionately high IMDb ratings relative to their revenue — suggesting that critical success and financial success are not always strongly aligned.

Among genres, we expect action, adventure, and fantasy films to show the strongest positive budget-revenue correlation, since these genres depend most heavily on spectacle and wide theatrical distribution to justify their investment, while drama and horror — genres that can succeed critically and commercially on low budgets — will show weaker or more variable correlations.

```python
# Setup — run once after cloning to download raw data
# %pip install requests tqdm
import sys
sys.path.append('./modules')
import get_data

datafiles = [
    { 'url': 'https://raw.githubusercontent.com/danielgrijalvas/movies/master/movies.csv',
      'filename': 'movies.csv' },
    { 'url': 'https://raw.githubusercontent.com/utkarshx27/movies-dataset/main/movie_dataset.csv',
      'filename': 'movie_dataset.csv' }
]
get_data.get_raw(datafiles, destination_directory='data/00-raw/')
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import statsmodels.formula.api as smf
from scipy import stats

sns.set_style('whitegrid')
plt.rcParams['figure.dpi'] = 100

# ── Load and clean Dataset 1 ────────────────────────────────────────────────
df = pd.read_csv('data/00-raw/movies.csv')
print('Shape:', df.shape)
df.head()

df2 = pd.read_csv('data/00-raw/movie_dataset.csv')


# Missingness summary
missing_percent = (df.isnull().mean() * 100).sort_values(ascending=False)
missing_table = pd.DataFrame({
    'Missing Values': df.isnull().sum(),
    'Percent Missing': missing_percent
})
missing_table.sort_values(by='Percent Missing', ascending=False)

# Parse release date, drop post-release variable
df['release_date'] = pd.to_datetime(
    df['released'].str.extract(r'^([^(]+)').squeeze().str.strip(), errors='coerce'
)
df = df.drop(columns=['released', 'votes'])

str_cols = df.select_dtypes(include='object').columns
df[str_cols] = df[str_cols].apply(lambda s: s.str.strip())

for col in ['budget', 'gross', 'runtime']:
    df[col] = pd.to_numeric(df[col], errors='coerce')

df.loc[df['budget'] == 0, 'budget'] = np.nan
df.loc[df['gross'] == 0, 'gross'] = np.nan

print('After cleaning, shape:', df.shape)
print('Missing budget:', df['budget'].isnull().sum())
print('Missing gross:', df['gross'].isnull().sum())

df = df.dropna(subset=['budget'])
print('After dropping missing budget:', df.shape)   

# CPI-U inflation adjustment to 2020 dollars
cpi = {
    1980: 82.4, 1981: 90.9, 1982: 96.5, 1983: 99.6, 1984: 103.9,
    1985: 107.6, 1986: 109.6, 1987: 113.6, 1988: 118.3, 1989: 124.0,
    1990: 130.7, 1991: 136.2, 1992: 140.3, 1993: 144.5, 1994: 148.2,
    1995: 152.4, 1996: 156.9, 1997: 160.5, 1998: 163.0, 1999: 166.6,
    2000: 172.2, 2001: 177.1, 2002: 179.9, 2003: 184.0, 2004: 188.9,
    2005: 195.3, 2006: 201.6, 2007: 207.3, 2008: 215.3, 2009: 214.5,
    2010: 218.1, 2011: 224.9, 2012: 229.6, 2013: 233.0, 2014: 236.7,
    2015: 237.0, 2016: 240.0, 2017: 245.1, 2018: 251.1, 2019: 255.7,
    2020: 258.8
}
BASE_CPI = cpi[2020]
df['cpi'] = df['year'].map(cpi)
df['budget_real'] = df['budget'] * (BASE_CPI / df['cpi'])
df['gross_real'] = df['gross'] * (BASE_CPI / df['cpi'])
df = df.drop(columns=['cpi'])

df['log_budget'] = np.log1p(df['budget_real'])
df['log_gross'] = np.log1p(df['gross_real'])
df['era'] = df['year'].apply(lambda y: 'Post-2015' if y >= 2015 else 'Pre-2015')
df['us_film'] = df['country'].apply(lambda c: 'US' if c == 'United States' else 'non-US')

print('Final Dataset 1 shape:', df.shape)
df[['budget_real','gross_real','score','runtime']].describe()


print('Movies by Era:')
print(df['era'].value_counts())
print('\nUS vs Non-US:')
print(df['us_film'].value_counts())

df.reset_index(drop=True).to_csv('data/02-processed/movies_clean.csv', index=False)
print('\nDataset 1 saved.')
```

## Merging the Datasets 

We merge on movie title and release year after standardizing column names. An inner join is used to produce a combined dataset with full information from both sources. Unmatched rows are noted for reference. After merging we add log-transformed financial columns and the era/US indicator columns needed for EDA.

```python

df1 = pd.read_csv('data/00-raw/movies.csv')
df2 = pd.read_csv('data/00-raw/movie_dataset.csv')
print(f'Dataset 1: {df1.shape}')
print(f'Dataset 2: {df2.shape}')

df1['merge_title'] = df1['name'].str.lower().str.strip()
df1['merge_year'] = df1['year']
df2['merge_title'] = df2['title'].str.lower().str.strip()
df2['merge_year'] = pd.to_datetime(df2['release_date'], errors='coerce').dt.year

df1_clean = df1.dropna(subset=['merge_title', 'merge_year']).copy()
df2_clean = df2.dropna(subset=['merge_title', 'merge_year']).copy()
df2_clean['merge_year'] = df2_clean['merge_year'].astype(int)

df1_cols = ['merge_title', 'merge_year', 'name', 'year', 'genre', 'rating',
            'score', 'votes', 'director', 'star', 'country', 'budget',
            'gross', 'company', 'runtime']
df2_cols = ['merge_title', 'merge_year', 'genres', 'original_language',
            'production_countries', 'popularity', 'revenue', 'vote_average',
            'vote_count', 'cast']

merged = pd.merge(
    df1_clean[df1_cols], df2_clean[df2_cols],
    on=['merge_title', 'merge_year'], how='inner', suffixes=('_ds1', '_ds2')
)
merged = merged.drop(columns=['merge_title', 'merge_year'])

# Derived columns
merged['log_budget'] = np.log1p(merged['budget'])
merged['log_gross'] = np.log1p(merged['gross'])
merged['era'] = merged['year'].apply(lambda y: 'Post-2015' if y >= 2015 else 'Pre-2015')
merged['is_us'] = merged['country'] == 'United States'

print(f'\nMerged dataset: {merged.shape}')
print(f'Columns: {merged.columns.tolist()}')
merged.to_csv('data/02-processed/merged_movies.csv', index=False)
merged.head()
```

#### Section 1: Univariate Distributions of Key Numeric Variables

We start by examining the distributions of our three core metrics: production budget, gross revenue, and IMDb score. Budget and gross are both heavily right-skewed — a small number of blockbusters earn orders of magnitude more than typical films — which is why we plot them on a log scale. The log-transformed distributions are approximately bell-shaped, confirming that a log-normal model is appropriate for these variables. IMDb score is roughly normally distributed and centered around 6.5–7.0, consistent with prior literature. The median budget ($30M) is far lower than the mean, further confirming the right skew and justifying our use of medians for group comparisons.

**Graph 1 — Univariate Distributions (Budget, Revenue, IMDb Score)**

Both budget and revenue follow approximately log-normal distributions, with peaks around $30M for budget and $55M for gross. Revenue is considerably more variable than budget — its distribution spans roughly 6 log-scale bins versus 4 for budget — meaning studios spend within a relatively predictable range but earn back widely varying amounts. A small bump of very-low-budget films indicates the presence of indie productions. IMDb scores are approximately normally distributed and centered at 6.6.

**Low-budget outlier check:** Films with budgets under $100K include notable exceptions like Paranormal Activity ($15K → $193M gross) and The Blair Witch Project ($60K → $248M). These represent a small but distinct subpopulation with extreme ROIs and will appear as influential outliers in regression models.

**Graph 10-  OLS Regression**

We fit an OLS regression model predicting log gross revenue from log budget, US vs. non-US indicator, era, their interactions, genre fixed effects, and IMDb score. The model explains approximately 50% of the variance in log gross (R² ≈ 0.507). Log budget is the dominant predictor (coef ≈ 0.87, p < 0.001), confirming the budget-revenue relationship holds after controlling for all other factors. The interaction terms for US status and era are not statistically significant at conventional levels, suggesting that while raw correlations differ between groups, the budget-revenue slope does not differ significantly after controlling for genre and score. IMDb score has a small but statistically significant positive effect on gross (coef ≈ 0.20, p < 0.001), meaning better-reviewed films do earn slightly more — but the effect is far smaller than that of budget.

```python

# Graph 10 OLS Regression
df_reg = df.dropna(subset=['log_budget','log_gross','score','genre','is_us','era'])

model = smf.ols(
    """
    log_gross ~ log_budget
              + C(is_us)
              + C(era)
              + log_budget:C(is_us)
              + log_budget:C(era)
              + C(genre)
              + score
    """,
    data=df_reg
).fit()

print(model.summary())
```

```python
# Residuals vs Fitted plot
fig, ax = plt.subplots(figsize=(7, 6))
ax.scatter(model.fittedvalues, model.resid, alpha=0.4, s=15, color='steelblue')
ax.axhline(0, color='red', linewidth=1.2, linestyle='--')
ax.set_xlabel('Fitted log(Gross)')
ax.set_ylabel('Residuals')
ax.set_title('OLS Residuals vs. Fitted Values')
plt.tight_layout()
plt.savefig('results/fig11_ols_residuals.png', dpi=100, bbox_inches='tight')
plt.show()

print(f'R-squared:       {model.rsquared:.3f}')
print(f'Adj. R-squared:  {model.rsquared_adj:.3f}')
print(f'N observations:  {int(model.nobs)}')
```

**Regression Results & Interpretation**

The OLS model explains approximately 50% of the variance in log gross revenue (R² ≈ 0.50), strong for a cross-sectional model of film performance.

Key findings:
- **Log budget** is the dominant predictor (coef ≈ 0.87, p < 0.001). A 1% increase in production budget is associated with roughly a 0.87% increase in gross revenue, holding all else constant.
- **US films** earn significantly more than non-US films (p < 0.05), consistent with Hollywood's distribution advantage. The interaction term (log_budget × US) is not significant — the *slope* does not differ by country once other factors are controlled.
- **Era:** Neither the era main effect nor its interaction is statistically significant after controlling for genre and budget, suggesting the apparent post-2015 strengthening observed in EDA reflects composition differences rather than a true era effect.
- **IMDb score** has a small but significant positive effect (coef ≈ 0.20, p < 0.001): better-reviewed films earn slightly more, but far less than budget explains.
- The residual plot shows no major systematic pattern, supporting the appropriateness of a linear model in log space.

Overall, after conducting our analysis, we were able to support our hypothesis that the existing factors that impact a movie's success. Our results showed that higher production budgets had a large influence on the financial success of films, confirmed both by the strong log-log correlation (r ≈ 0.67) in EDA and by the OLS regression coefficient (≈ 0.87, p < 0.001) which held even after controlling for genre, country, era, and IMDb score. Lastly, post-2015 films showed a slightly stronger budget-revenue relationship rather than a weaker one as hypothesized, which we attribute to a selection effect: lower-budget films increasingly migrated to streaming platforms post-2015, leaving theatrical data dominated by high-budget wide releases and mechanically strengthening the observed correlation.

The film industry is an ever-changing phenomenon that drives popular culture. The definition of a "successful" film cannot be explained by a single indicator. Financial success was most strongly associated with production-related characteristics such as budget, genre, release period, and country of origin. Success in film is not measured solely by revenue — many movies that underperform financially still leave an enduring mark on audiences and popular culture. The modern film industry is highly complex and constantly evolving, reflecting current economic and political conditions of society.

If we were to continue this project, we would like to include more international films and incorporate streaming performance data to better capture how movies generate revenue today.
